```python
from pyspark.sql.functions import col, lower
df1=spark.read.json("/Volumes/prodcatalogdef/prodschemadef/datalake/logistics_shipment_detail_3000.json",multiLine=True,mode="PERMISSIVE")

df1.filter(
    (lower(col("vehicle_type"))== "truck") &
    (lower(col("shipment_date"))>= "24-10-01") &
    (lower(col("destination_city"))== "chennai"))\
        .write.mode("overwrite")\
        .format("delta")\
        .saveAsTable("lakeflow_pl_cat.lakeflow_pl_sch.shipments")

df2 = spark.sql("select * from lakeflow_pl_cat.lakeflow_pl_sch.shipments")

display(df2)
```
